```python
# 4. ROLL D : Visualiseerimine
# 3 Plotly diagrammi (segmentide jaotus, hajuvusdiagramm, top VIP)
# Graafik 1: Segmentide jaotus
import plotly.express as px
segment_counts = rfm['segment'].value_counts().reset_index()
segment_counts.columns = ["Segment", "Klientide_arv"]
fig1 = px.bar(
    segment_counts,
    x="Segment",
    y="Klientide_arv",
    title='UrbanStyle: Kliendisegmentide jaotus (RFM)',
    text="Klientide_arv"
    #labels={'segment': 'Segment', 'count': 'Klientide arv'},
    #color='segment'
)
fig1.show()
#fig1.write_html("segmentide_klientide_arv.html")

# Graafik 2 : Hajuvusdiagramm (Recency vs Monetary väärtus segmentide kaupa)
fig2 = px.scatter(
    rfm,
    x="Recency",
    y="Monetary",
    color="segment",
    size="Frequency",
    hover_data=["customer_id"],
    title="Recency vs Monetary väärtus segmentide kaupa",
    labels={
        "Recency": "Päevi viimasest ostust",
        "Monetary": "Kogukulu",
        "Frequency": "Ostude arv"
    }
)

fig2.show()
#fig2.write_html("hajuvusdiagramm.html")

# Graafi 3: Top 10 VIP klienti 
vip_top10 = (
    rfm[rfm["segment"] == "VIP Champions"]
    .nlargest(10, "Monetary")
    .sort_values("Monetary")
)

fig3 = px.bar(
    vip_top10,
    x="Monetary",
    y="customer_id",
    orientation="h",
    title="Top 10 VIP klienti kogukulutuse järgi",
    text="Monetary"
)

fig3.update_layout(
    xaxis_title="Kogukulu",
    yaxis_title="Kliendi ID"
)

fig3.show()
#fig3.write_html("VIP_klientide_top10.html")
```
